Possibilidades do que fazer com os dados nulos (.isna() ou .isnull() e fazendo um .sum()), no geral:

# Possibilidades:
1. Remover os dados;
2. Preencher os numéricos com: média, mediana, 999 ou -999
3. Para preencher booleanos: se aquele dado não existe, ver se faz sentido, colocar ele como um "não"/0
4. Para preencher textos: "Desconhecido", "Não informado"

# Escolha do que fazer.

- Entrega_doc_2: nós podemos optar em colocar como 'N'
- valores das colunas de score: nesse momento, podemos optar pela mediana
- pais: podemos colocar como Desconhecido

```python
#Corrigindo dados dos paises

df_unificado['pais'] = df_unificado['pais'].fillna('Deconhecido')

print(df_unificado['pais'].isna().value_counts())
print(df_unificado['pais'].value_counts())
```

```python
#Corrigindo dados dos documentos_entrega2

df_unificado['entrega_documento2'] = df_unificado['entrega_documento2'].fillna("N")

df_unificado['entrega_documento2'].isna().sum()
df_unificado['entrega_documento2'].value_counts()
```

```python
#Preenchendo os scores que estão vazios/nulos

df_unificado.isna().sum()

scores_nulos = ['score_2', 'score_3', 'score_4', 'score_6', 'score_9', 'score_10']

for score in scores_nulos:
    mediana = df_unificado[score].median()
    df_unificado[score] = df_unificado[score].fillna(mediana)
    print(df_unificado[score].isna().sum())
```

Nas entregas do documento, substituir y e n por 0 e 1

```python
df_unificado['entrega_documento2'] = df_unificado['entrega_documento2'].map({"Y":1, "N": 0})
df_unificado['entrega_documento3'] = df_unificado['entrega_documento3'].map({"Y":1, "N": 0})

print(df_unificado['entrega_documento1'].value_counts())
print(df_unificado['entrega_documento2'].value_counts())
print(df_unificado['entrega_documento3'].value_counts())
```

AGRUPAMENTO POR PAÍSES

Paises com mais de 10k de transações, agrupamento individual
Menos de 10k em um mesmo agrupamento

```python
def agrupar_paises(pais):
    if pais in ["BR", "AR"]:
        return pais
    else:
        return "Outros"
```

```python
df_unificado['pais_agrupado'] = df_unificado['pais'].apply(agrupar_paises)
df_unificado['pais_agrupado'].value_counts()
```

```python
#aplicando função com lambda

paises_agrupados = lambda pais: pais if pais in ["BR" , "AR"] else "Outros"

df_unificado['pais_agrupado'] = df_unificado['pais'].apply(paises_agrupados)
df_unificado['pais_agrupado'].value_counts()
```

A partir da data de compra, criar a hora e o dia da semana da compra

```python
dias_semana_dict = {'6': "Domingo",
                    '0': "Segunda",
                    '1': "Terça",
                    '2': "Quarta",
                    '3': "Quinta",
                    '4': "Sexta",
                    '5': "Sábado"}
```

```python
df_unificado['dia_semana'] = df_unificado['data_compra'].dt.dayofweek.astype(str).map(dias_semana_dict)
```

```python
df_unificado['dia_semana'].unique()
```

```python
df_unificado["hora_compra"] = df_unificado["data_compra"].dt.strftime("%H:%M")

df_unificado['hora_compra'].head()
```
